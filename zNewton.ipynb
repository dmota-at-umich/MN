```python
import numpy as np
```

```python
def zNewton(f, g, x0, itermax, tol):
    """
    Implementación del método de Newton.
    """
    # Aplanar x0 a vector unidimensional (1D)
    x0 = np.asarray(x0, dtype=float).flatten()
    flag = True
    orbita = [x0.copy()]
    iter_count = 1
    
    xk = x0
    error = 0.0

    while flag and iter_count < itermax:
        fx = np.asarray(f(x0), dtype=float)
        gx = np.asarray(g(x0), dtype=float)

        # Cálculo del paso sk = -g(x0) \ f(x0)
        if gx.ndim == 0 or gx.size == 1:
            sk = -fx / gx
        else:
            sk = np.linalg.solve(gx, -fx)

        xk = x0 + sk
        error = float(np.linalg.norm(xk - x0))

        norm_x0 = float(np.linalg.norm(x0))
        if norm_x0 > 0 and error < tol * norm_x0:
            flag = False  # Criterio de paro relativo
        elif norm_x0 == 0 and error < tol:
            flag = False  # Criterio de paro absoluto si x0 es cero
            
        orbita.append(xk.copy())
        iter_count += 1
        x0 = xk

    return xk, orbita, error
```

```python

f = lambda X: np.array([X[0]**2 + X[1]**2 - 1, X[0] - X[1]])
g = lambda X: np.array([[2 * X[0], 2 * X[1]], [1, -1]])
```

```python
x0 = np.array([0.5, 0.5])
xk, orbita, error = zNewton(f, g, x0, 1e3, 1e-6)

print("Solución encontrada:", xk)
print("Error final:", error)
```
