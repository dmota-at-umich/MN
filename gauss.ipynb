```python
import numpy as np
```

```python
def thebest(V):
    """Encuentra el índice del elemento con el máximo valor absoluto en un vector."""
    V = np.abs(V.flatten())
    mu = 0
    maxim = V[0]
    for i in range(1, len(V)):
        if V[i] > maxim:
            mu = i
            maxim = V[i]
    return mu
```

```python
def gauss(A):
    """Eliminación Gaussiana simple con pivoteo parcial."""
    A = np.array(A, dtype=float, copy=True)
    n = A.shape[0]
    p = np.arange(n)
    epsilon = np.sqrt(np.finfo(float).eps)

    for k in range(n - 1):
        # Encuentra el pivote local y ajusta al índice global
        mu = thebest(A[k:n, k]) + k

        if mu != k:
            # Intercambia filas en A desde la columna k en adelante
            A[[k, mu], k:] = A[[mu, k], k:]
            # Intercambia índices en el vector de permutación
            p[k], p[mu] = p[mu], p[k]

        if abs(A[k, k]) > epsilon:
            # Factor de eliminación
            tau = A[k + 1 : n, k] / A[k, k]
            # Eliminación en la submatriz
            A[k + 1 : n, :] -= tau[:, np.newaxis] * A[k, :]

    return A, p
```

```python
def backward(U):
    """Sustitución regresiva (versión fila).

    Recibe una matriz aumentada U de tamaño (n, n+1).
    """
    U = np.array(U, dtype=float)
    n = U.shape[0]
    b = U[:, n].copy()

    b[n - 1] = b[n - 1] / U[n - 1, n - 1]

    for i in range(n - 2, -1, -1):
        b[i] = (b[i] - np.dot(U[i, i + 1 : n], b[i + 1 : n])) / U[i, i]

    return b
```

```python
A=np.array([[3,2],[2,6]])
b=np.array([[5],[8]])
```

```python
U,_ = gauss(np.hstack((A,b)))
x = backward(U)
print("Soluci'on: ",x)
```
