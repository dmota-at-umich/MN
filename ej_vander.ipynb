```python
import numpy as np
#from scipy.interpolate import vander
from scipy import interpolate

import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1,1,11)
y = 1./(1+25*x**2)
```

```python
C=np.vander(x)
P=np.linalg.solve(C,y)
y1d = interpolate.interp1d(x, y)
```

```python
P
```

```python
x2 = np.linspace(-1,1,1001)
van = np.polyval(P, x2)


plt.scatter(x,y)
plt.plot(x2,y1d(x2),label=r"Lineal", linestyle='-')
plt.plot(x2,van,label=r"VanderMonde", linestyle='-.')

plt.legend()
plt.grid()
plt.show()
```
